## Helper Functions

```python
def display_content(base_path, patient_id):
    patient_folder = f"patient{patient_id:03d}"
    full_path = os.path.join(base_path, patient_folder)
    
    if os.path.exists(full_path):
        files = os.listdir(full_path)
        print(f"Contents of {patient_folder}:")
        for f in files:
            print(f" - {f}")
    else:
        print(f"Error: Path not found - {full_path}")

def read_info_cfg(base_path, patient_id):
    patient_folder = f"patient{patient_id:03d}"
    full_path = os.path.join(base_path, patient_folder)

    if os.path.exists(full_path):
        files = os.listdir(full_path)
        
        cfg_files = [f for f in files if f.endswith('.cfg')]
        
        if not cfg_files:
            print(f"No .cfg files found in {patient_folder}")
            return

        for filename in cfg_files:
            file_path = os.path.join(full_path, filename)
            try:
                with open(file_path, 'r') as f:
                    content = f.read()
                
                print(f"--- CONFIG: {filename} ---")
                print(content.strip())
                
            except Exception as e:
                print(f"Error reading {filename}: {e}")
    else:
        print(f"Error: Path not found - {full_path}")

def read_md(base_path, patient_id):
    patient_folder = f"patient{patient_id:03d}"
    full_path = os.path.join(base_path, patient_folder)

    if os.path.exists(full_path):
        files = os.listdir(full_path)
        
        cfg_files = [f for f in files if f.endswith('.md')]
        
        if not cfg_files:
            print(f"No .md files found in {patient_folder}")
            return

        for filename in cfg_files:
            file_path = os.path.join(full_path, filename)
            try:
                with open(file_path, 'r') as f:
                    content = f.read()
                
                print(f"--- MD: {filename} ---")
                print(content.strip())
                
            except Exception as e:
                print(f"Error reading {filename}: {e}")
    else:
        print(f"Error: Path not found - {full_path}")
```
